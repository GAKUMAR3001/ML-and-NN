The R-squared (R²) score measures the proportion of the variance in the dependent variable that is predictable from the independent variables. Interestingly, when you add more features to a regression model, the R² score can increase, even if those additional features are not truly related to the outcome. This phenomenon occurs due to how R² is calculated and the nature of regression models. Here's why this happens and the intuition behind it:

### R² Calculation and Overfitting

- **R² and Model Complexity**: The R² score is calculated based on the proportion of total variance explained by the model. When you add more variables (features) to a model, you are inherently increasing the complexity of the model. By doing so, the model may start to capture more of the variance in the training data — including the noise — which can artificially inflate the R² value.

- **Fitting the Noise**: Adding more features to a model, regardless of their relevance, provides the model with more opportunities to fit the data more closely. This includes fitting the noise in the data, which is not generalizable beyond the training set. While the model appears to perform better on the training data (indicated by a higher R²), it may not actually improve the model's predictive power on unseen data.

### Mathematical Perspective

- **Degrees of Freedom and Overfitting**: Each additional feature introduces more parameters for the model to estimate. As the number of parameters increases, the model has more "flexibility" to fit the data closely. Mathematically, this reduces the sum of squares of residuals (the differences between observed and predicted values), which directly increases R². However, fitting more closely to the training data can mean capturing noise, leading to overfitting.

### Intuition Behind R² Increase

- **Capturing Variance vs. Predictive Power**: The essence of the increase in R² with additional features is that the model is capturing more of the variance in the training data. However, capturing more variance does not necessarily mean improving the model's accuracy or its ability to generalize. Some of this additional variance captured might just be random noise or patterns that do not exist in the broader dataset.

### Adjusted R²

To mitigate this issue and get a more accurate measure of model performance, especially when adding more features, the adjusted R² metric is used. Adjusted R² takes into account the number of predictors (features) in the model relative to the number of observations, adjusting the R² value based on the number of features and penalizing for adding features that do not improve model performance on new data. This provides a more nuanced assessment of model quality, especially in the context of model complexity and the risk of overfitting.

In summary, the increase in R² when adding irrelevant features is a reminder of the importance of careful feature selection and the need to use more comprehensive metrics, like adjusted R², to evaluate model performance accurately.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

!gdown 1UpLnYA48Vy_lGUMMLG-uQE1gf_Je12Lh
```

```python
df = pd.read_csv('cars24-car-price-clean.csv')
df.head()
```

```python
y=df[['selling_price']]
X=df.drop('selling_price', axis=1)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
y_train = np.array(y_train)
```

```python
X_sm = sm.add_constant(X_train)
X_sm
```

```python
model = sm.OLS(y_train, X_sm)
results = model.fit()

# print the summary

print(results.summary())
```

```python
output = results.predict(X_sm)
```
